Normalisation

```python
import pandas as pd
import ast

def normalize_text_column(col):
    def clean(val):

        if pd.isna(val):
            return "Unknown"


        if isinstance(val, list):
            return ", ".join([str(v).strip() for v in val if v and str(v).strip() != ""]) or "Unknown"
        if isinstance(val, str):
            try:
                parsed = ast.literal_eval(val)
                if isinstance(parsed, list):
                    parsed_clean = [str(v).strip() for v in parsed if v and str(v).strip() != ""]
                    return ", ".join(parsed_clean) if parsed_clean else "Unknown"
                else:
                    return str(parsed).strip() or "Unknown"
            except:
                return val.strip() if val.strip() else "Unknown"

        # 4️⃣ Cas par défaut
        return "Unknown"

    return col.apply(clean)

# ✅ Application sur tes colonnes
sbooks_data_df['authors'] = normalize_text_column(sbooks_data_df['authors'])
sbooks_data_df['categories'] = normalize_text_column(sbooks_data_df['categories'])
```

```python
from pyspark.sql.functions import col, avg, count, length, explode, split, lower, regexp_replace,when, col, udf
```

```python
sbooks_rating_df = ratings_df.select("Title", "review/score", "review/text")
sbooks_rating_df = sbooks_rating_df.fillna({
    "Title": "Unknown",
    "review/score": 0,
    "review/text": "Unknown"
})

print("=== Null values in sbooks_rating_df ===")
sbooks_rating_df.select([count(when(col(c).isNull(), c)).alias(c) for c in sbooks_rating_df.columns]).show()

sbooks_data_df = books_df.select("Title", "authors", "publisher", "categories", "ratingsCount")

sbooks_data_df = sbooks_data_df.fillna({
    "Title": "Unknown",
    "authors": "Unknown",
    "publisher": "Unknown",
    "categories": "Unknown",
    "ratingsCount": 0
})

print("=== Null values in sbooks_data_df ===")
sbooks_data_df.select([count(when(col(c).isNull(), c)).alias(c) for c in sbooks_data_df.columns]).show()
```

```python
from pyspark.sql.types import StringType
import ast

def normalize_list(x):
    try:
        if isinstance(x, str):
            x = ast.literal_eval(x)
        # Si c'est une liste et non vide, prendre le premier élément
        if isinstance(x, list) and len(x) > 0:
            return str(x[0])
        else:
            return "Unknown"
    except (ValueError, SyntaxError, TypeError):
        return "Unknown"

normalize_list_udf = udf(normalize_list, StringType())

sbooks_data_df = sbooks_data_df.withColumn("authors", normalize_list_udf(col("authors")))
sbooks_data_df = sbooks_data_df.withColumn("categories", normalize_list_udf(col("categories")))

sbooks_data_df.select("Title", "authors", "categories").show(10, truncate=False)
```

```python
book_df = sbooks_data_df.join(
    sbooks_rating_df,
    on="Title",
    how="inner"  # options: 'inner', 'left', 'right', 'outer'
)

print("=== Colonnes après join ===")
print(book_df.columns)

print("=== Aperçu du DataFrame ===")
book_df.show(5, truncate=False)
```
