```python
import ast

# Function to safely convert string representation of list to a list
def safe_literal_eval(x):
    try:
        return ast.literal_eval(x)
    except (ValueError, SyntaxError):
        return []

# Apply the function to 'authors' and 'categories' columns in sbooks_data_df
sbooks_data_df['authors'] = sbooks_data_df['authors'].apply(safe_literal_eval)
sbooks_data_df['categories'] = sbooks_data_df['categories'].apply(safe_literal_eval)

print("Normalized 'authors' and 'categories' columns in sbooks_data_df:")
display(sbooks_data_df[['Title', 'authors', 'categories']].head())
```

```python
from collections import Counter
import re

# Initialize the tracker with detailed logging
tracker = EmissionsTracker(
    project_name="demo_experiment",
    output_dir=log_dir,
    measure_power_secs=1,        # frequency of power measurement
    log_level="info",            # ensures detailed console + file logs
    save_to_file=True,           # ensure CSV file is written
)

# Start tracking
tracker.start()

# Simulate a computation workload
for _ in range(1000):
    a = np.random.rand(500, 500)
    b = np.random.rand(500, 500)
    np.dot(a, b)


# Fusionner tous les textes en une seule chaîne
all_text = " ".join(book_df['review/text'].dropna().astype(str))

# Nettoyer le texte : mettre en minuscule, retirer ponctuation
all_text = re.sub(r'[^\w\s]', '', all_text.lower())

# Créer une liste de mots
words = all_text.split()

# Compter la fréquence de chaque mot
word_counts = Counter(words)

# Top 20 mots les plus fréquents
most_common_words = word_counts.most_common(20)

print("Top 20 most frequent keywords:")
for word, count in most_common_words:
    print(f"{word}: {count}")

# Stop tracking
emissions = tracker.stop()

# Print summary
print("\n=== CodeCarbon Tracking Summary ===")
print(f"Project: {tracker._project_name}")
print(f"Output CSV: {os.path.join(log_dir, 'emissions_pandas.csv')}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import drive
drive.mount('/gdrive')

```

```python
parquet_path = "/gdrive/MyDrive/TP2_Green_IT/Data/carbon_logs_TP3/emission_parquet.csv"
emission_parquet = pd.read_csv(parquet_path,sep=';')

csv_path = "/gdrive/MyDrive/TP2_Green_IT/Data/carbon_logs_TP3/emission_csv.csv"
emission_csv = pd.read_csv(csv_path,sep=';')

emission_csv['Format'] = 'CSV'
emission_parquet['Format'] = 'Parquet'
emission_parquet.columns = emission_parquet.columns.str.strip()
emission_csv.columns = emission_csv.columns.str.strip()
```

```python
emission_csv['CO₂ (kg)'] = emission_csv['CO₂ (kg)'] * 1000   # g
emission_parquet['CO₂ (kg)'] = emission_parquet['CO₂ (kg)'] * 1000  # g
```

```python
emissions_df = pd.concat([emission_csv, emission_parquet], ignore_index=True)
```

```python
emissions_df['CO₂ (kg)'] = pd.to_numeric(emissions_df['CO₂ (kg)'], errors='coerce') * 1000

# 🔹 Agréger CO₂ total par format
co2_summary = emissions_df.groupby('Format')['CO₂ (kg)'].sum().reset_index()

# 🔹 Plot
plt.figure(figsize=(8,5))
sns.barplot(x='Format', y='CO₂ (kg)', data=co2_summary, palette='salmon')
plt.title("Émissions totales de CO₂ par format")
plt.ylabel("CO₂ émis (g)")
plt.xlabel("Format")
plt.tight_layout()
plt.show()
```
